```python
dataset_path="../datasets/ArchiStyles/"
```

```python
import torch
from torchvision import datasets, transforms
from torch.utils.data import DataLoader
import clip


device = "cuda" if torch.cuda.is_available() else "cpu"
print(device)
```

```python

model, preprocess = clip.load("ViT-B/32", device=device)

#image = preprocess(Image.open("../datasets/AechiStyles/valid/Achaemenid/000016_jpg.rf.957ece026984deff230bb96e077acc5b.jpg")).unsqueeze(0).to(device)

dataset = datasets.ImageFolder(dataset_path, transform=preprocess)
loader = DataLoader(dataset, batch_size=32, num_workers=4,)
text = clip.tokenize(['Achaemenid', 'American Foursquare', 'American craftsman style', 'Ancient Egyptian', 'Art Deco', 'Art Nouveau', 'Baroque', 'Bauhaus', 'Beaux-Arts', 'Byzantine', 'Chicago school', 'Colonial', 'Deconstructivism', 'Edwardian', 'Georgian', 'Gothic', 'Greek Revival', 'International style', 'Novelty', 'Palladian', 'Postmodern', 'Queen Anne', 'Romanesque', 'Russian Revival', 'Tudor Revival']).to(device)
with torch.no_grad():
    text_features = model.encode_text(text)
    total=0
    true_positives=0
    for images, labels in loader:
        images = images.to(device)
        labels=labels.to(device)
        # features = model.encode_image(images)
        
        logits_per_image, logits_per_texts = model(images, text)
        # probs = logits_per_image.softmax(dim=-1).cpu().numpy()
        true_positives += (torch.argmax(logits_per_image)==labels).sum() # sum of all positive
        total+=1
    
    accuracy= true_positives / total


print("the accuracy is ",accuracy.values)
```
