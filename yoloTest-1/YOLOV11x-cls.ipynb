```python
from ultralytics import YOLO
import numpy as np
```

### Hyperparams

```python
dataset="../datasets/ArchiStyles"
imgsz=224 # The original image size for the ArchiStyle dataset is 620x640 
epochs=100
```

## Training

```python
model = YOLO("yolo11x-cls.pt")  # load a pretrained model (recommended for training)

# Train the model
results = model.train(data=dataset, epochs=epochs, imgsz=imgsz,device=-1,cache='ram',workers=10)
model.export()
```

```python
model.export()
```

## Validation

```python
metrics = model.val() 
```
